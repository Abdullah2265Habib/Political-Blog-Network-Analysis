### Loading the Dataset & Resolving the NetworkX GML Duplication Issue

```python
with open(RAW_DATA_PATH, "r", encoding="utf-8") as f: raw_content = f.read()

multi_content = raw_content.replace("graph [\n  directed 1", "graph [\n  directed 1\n  multigraph 1")
G_raw = nx.parse_gml(multi_content, label=None)

print("Raw Graph Data: ")
print(f"graph Class:              {type(G_raw).__name__}")
print(f"is Directed:              {G_raw.is_directed()}")
print(f"is Multigraph:            {G_raw.is_multigraph()}")
print(f"raw Node Count (|V|):     {G_raw.number_of_nodes():,}")
print(f"raw Edge Count (|E|):     {G_raw.number_of_edges():,}")
```

### Preprocessing the Audit Log

```python
missing_label = sum(1 for _, d in G_raw.nodes(data=True) if "label" not in d)
missing_val   = sum(1 for _, d in G_raw.nodes(data=True) if "value" not in d)
missing_src   = sum(1 for _, d in G_raw.nodes(data=True) if "source" not in d)
raw_val_dist  = Counter(d.get("value") for _, d in G_raw.nodes(data=True))

raw_self_loops = list(nx.selfloop_edges(G_raw))

edge_pair_counts = Counter((u, v) for u, v, _ in G_raw.edges(keys=True))
multi_pairs = {pair: count for pair, count in edge_pair_counts.items() if count > 1}
total_excess_edges = sum(c - 1 for c in multi_pairs.values())

raw_isolates = list(nx.isolates(G_raw))
raw_wccs = list(nx.weakly_connected_components(G_raw))
raw_sccs = list(nx.strongly_connected_components(G_raw))
wcc_sizes = sorted([len(c) for c in raw_wccs], reverse=True)
```

```python
print("Raw Preprocessing Audit")
print(f"Missing Node Labels:            {missing_label} (100% complete)")
print(f"Missing Political Values:       {missing_val} (Distribution: {dict(raw_val_dist)})")
print(f"Missing Source Attributes:      {missing_src}")
print(f"Self-loops Identified:          {len(raw_self_loops)} edges")
print(f"Blog Pairs with Multi-links:    {len(multi_pairs)} (Excess duplicate edges: {total_excess_edges})")
print(f"Isolated Nodes (Degree = 0):    {len(raw_isolates)} blogs")
print(f"Total Weakly Connected Comps:   {len(raw_wccs)} (Sizes: Giant={wcc_sizes[0]}, 2nd={wcc_sizes[1]}, Singletons={sum(1 for s in wcc_sizes if s == 1)})")
print(f"Total Strongly Connected Comps: {len(raw_sccs)} (Largest SCC size: {len(max(raw_sccs, key=len))})")
```

# Part B

### Page Rank and Betweeness Centrality

```python
pr_scores = nx.pagerank(G_clean, alpha=0.85, max_iter=200, tol=1e-6)

bc_scores = nx.betweenness_centrality(G_clean, normalized=True)

nx.set_node_attributes(G_clean, pr_scores, "pagerank")
nx.set_node_attributes(G_clean, bc_scores, "betweenness")

pr_sorted = sorted(pr_scores.items(), key=lambda x: x[1], reverse=True)
bc_sorted = sorted(bc_scores.items(), key=lambda x: x[1], reverse=True)

top10_pr_table = []
max_pr = pr_sorted[0][1]
for rank, (nid, raw_val) in enumerate(pr_sorted[:10], start=1):
    top10_pr_table.append({
        "Rank": rank,
        "Node ID": nid,
        "Blog URL / Label": G_clean.nodes[nid]["label"],
        "Raw PageRank": raw_val,
        "Norm Score (PR / Max)": raw_val / max_pr,
        "Ideology": "Liberal (0)" if G_clean.nodes[nid]["value"] == 0 else "Conservative (1)"
    })
top10_pr_df = pd.DataFrame(top10_pr_table)

top10_bc_table = []
max_bc = bc_sorted[0][1]
for rank, (nid, raw_val) in enumerate(bc_sorted[:10], start=1):
    top10_bc_table.append({
        "Rank": rank,
        "Node ID": nid,
        "Blog URL / Label": G_clean.nodes[nid]["label"],
        "Raw Betweenness": raw_val,
        "Norm Score (BC / Max)": raw_val / max_bc,
        "Ideology": "Liberal (0)" if G_clean.nodes[nid]["value"] == 0 else "Conservative (1)"
    })
top10_bc_df = pd.DataFrame(top10_bc_table)

print("Top 10 Rankings Computed Successfully.")
display(top10_bc_df)
```

### Ranking Comparison: Agreements, Disagreements, and Rank Correlations

```python
nodes_list = list(G_clean.nodes())
pr_vector = [pr_scores[n] for n in nodes_list]
bc_vector = [bc_scores[n] for n in nodes_list]

spear_rho, spear_p = spearmanr(pr_vector, bc_vector)
kend_tau, kend_p   = kendalltau(pr_vector, bc_vector)

top10_pr_set = set(n for n, _ in pr_sorted[:10])
top10_bc_set = set(n for n, _ in bc_sorted[:10])
top20_pr_set = set(n for n, _ in pr_sorted[:20])
top20_bc_set = set(n for n, _ in bc_sorted[:20])

overlap_10 = len(top10_pr_set & top10_bc_set)
jaccard_10 = overlap_10 / len(top10_pr_set | top10_bc_set)

overlap_20 = len(top20_pr_set & top20_bc_set)
jaccard_20 = overlap_20 / len(top20_pr_set | top20_bc_set)

comparison_metrics_df = pd.DataFrame({
    "Comparative Metric": [
        "Spearman Rank Correlation (rho)",
        "Spearman p-value",
        "Kendall Tau Correlation (tau)",
        "Kendall p-value",
        "Top-10 Overlap Count",
        "Top-10 Jaccard Similarity",
        "Top-20 Overlap Count",
        "Top-20 Jaccard Similarity"
    ],
    "Calculated Value": [
        f"{spear_rho:.4f}",
        f"{spear_p:.2e}",
        f"{kend_tau:.4f}",
        f"{kend_p:.2e}",
        f"{overlap_10} / 10 common blogs",
        f"{jaccard_10:.4f}",
        f"{overlap_20} / 20 common blogs",
        f"{jaccard_20:.4f}"
    ]
})

comparison_metrics_df
```

### Community Detection and Pairwise Evaluation

```python
def evaluate_graph_communities(G_sub, seed=42):
    G_u = G_sub.to_undirected()
    comms = nx.community.louvain_communities(G_u, seed=seed)
    comm_map = {n: cid for cid, c in enumerate(comms) for n in c}
    
    nodes = list(G_sub.nodes())
    N = len(nodes)
    Y = np.array([G_sub.nodes[n]["value"] for n in nodes])
    C = np.array([comm_map[n] for n in nodes])
    
    iu = np.triu_indices(N, k=1)
    p = C[iu[0]] == C[iu[1]]
    t = Y[iu[0]] == Y[iu[1]]
    
    tp = int(np.sum(p & t))
    tn = int(np.sum((~p) & (~t)))
    fp = int(np.sum(p & (~t)))
    fn = int(np.sum((~p) & t))
    
    total_pairs = len(iu[0])
    acc = (tp + tn) / total_pairs if total_pairs > 0 else np.nan
    prec = tp / (tp + fp) if (tp + fp) > 0 else np.nan
    rec = tp / (tp + fn) if (tp + fn) > 0 else np.nan
    f1 = (2 * prec * rec) / (prec + rec) if (prec + rec) > 0 else np.nan
    mod = nx.community.modularity(G_u, comms)
    
    wccs = list(nx.weakly_connected_components(G_sub))
    sccs = list(nx.strongly_connected_components(G_sub))
    lwcc_len = len(max(wccs, key=len)) if wccs else 0
    
    return {
        "Nodes": N,
        "Edges": G_sub.number_of_edges(),
        "WCC Count": len(wccs),
        "Giant WCC Size": lwcc_len,
        "SCC Count": len(sccs),
        "Detected Comms": len(comms),
        "Modularity (Q)": mod,
        "Accuracy": acc,
        "Precision": prec,
        "Recall": rec,
        "F1 Score": f1,
        "TP": tp,
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "Total Pairs": total_pairs
    }

eval_orig = evaluate_graph_communities(G_clean, seed=SEED)
eval_targ = evaluate_graph_communities(G_targeted, seed=SEED)
eval_rand = evaluate_graph_communities(G_random, seed=SEED)

experiment_results_df = pd.DataFrame([
    {"Experiment Graph": "Original Graph (G_clean)", **eval_orig},
    {"Experiment Graph": "Targeted Removal (Top 2% Betweenness)", **eval_targ},
    {"Experiment Graph": "Random Removal (25 Random Nodes)", **eval_rand}
])

display(experiment_results_df[[
    "Experiment Graph", "Nodes", "Edges", "WCC Count", "Giant WCC Size",
    "Detected Comms", "Modularity (Q)", "Accuracy", "Precision", "Recall", "F1 Score"
]].style.format({
    "Modularity (Q)": "{:.4f}",
    "Accuracy": "{:.4f}",
    "Precision": "{:.4f}",
    "Recall": "{:.4f}",
    "F1 Score": "{:.4f}"
}))
```

### Profiling Major Detected Communities

```python
community_profile_rows = []

for cid, c in enumerate(canonical_comms_sorted):
    c_nodes = list(c)
    size = len(c_nodes)
    if size < 5:
        continue
    
    val_counts = Counter(G_clean.nodes[n]["value"] for n in c_nodes)
    c_libs = val_counts.get(0, 0)
    c_cons = val_counts.get(1, 0)
    dominant_ideology = "Conservative (1)" if c_cons > c_libs else "Liberal (0)"
    purity = max(c_cons, c_libs) / size * 100
    
    mean_pr = np.mean([pr_scores[n] for n in c_nodes])
    mean_bc = np.mean([bc_scores[n] for n in c_nodes])
    
    top_c_blogs = sorted(c_nodes, key=lambda n: pr_scores[n], reverse=True)[:3]
    top_c_labels = ", ".join([G_clean.nodes[n]["label"] for n in top_c_blogs])
    
    community_profile_rows.append({
        "Comm ID": cid,
        "Size (|V_c|)": size,
        "Network Share": f"{size / N_clean * 100:.1f}%",
        "Liberal (0)": c_libs,
        "Conservative (1)": c_cons,
        "Dominant Wing": dominant_ideology,
        "Ideological Purity": f"{purity:.1f}%",
        "Mean PageRank": mean_pr,
        "Mean Betweenness": mean_bc,
        "Top Influential Blogs": top_c_labels
    })

comm_profile_df = pd.DataFrame(community_profile_rows)

print("TABLE 3: MAJOR COMMUNITY PROFILES")
display(comm_profile_df.style.format({
    "Mean PageRank": "{:.6f}",
    "Mean Betweenness": "{:.6f}"
}))
```

```python
def create_confusion_matrix(eval_result, title):
    cm = pd.DataFrame(
        [
            [eval_result["TN"], eval_result["FP"]],
            [eval_result["FN"], eval_result["TP"]]
        ],
        index=["Actual: 0", "Actual: 1"],
        columns=["Predicted: 0", "Predicted: 1"]
    )

    print(f"\n{title}")
    display(cm)


# Original Graph
create_confusion_matrix(
    eval_orig,
    "Confusion Matrix — Original Graph (G_clean)"
)

# Targeted Removal
create_confusion_matrix(
    eval_targ,
    "Confusion Matrix — Targeted Removal (Top 2% BC)"
)

# Random Removal
create_confusion_matrix(
    eval_rand,
    "Confusion Matrix — Random Removal (25 Random Nodes)"
)
```

```python
def plot_confusion_matrix(eval_result, title, filename):

    cm = np.array([
        [eval_result["TN"], eval_result["FP"]],
        [eval_result["FN"], eval_result["TP"]]
    ])

    fig, ax = plt.subplots(figsize=(7, 6))

    im = ax.imshow(cm)

    # Add values inside cells
    for i in range(2):
        for j in range(2):
            ax.text(
                j, i,
                f"{cm[i, j]:,}",
                ha="center",
                va="center",
                fontsize=14
            )

    ax.set_xticks([0, 1])
    ax.set_yticks([0, 1])

    ax.set_xticklabels([
        "Predicted 0",
        "Predicted 1"
    ])

    ax.set_yticklabels([
        "Actual 0",
        "Actual 1"
    ])

    ax.set_xlabel("Predicted")
    ax.set_ylabel("Actual")
    ax.set_title(title)

    plt.colorbar(im, ax=ax)

    plt.savefig(
        f"figures/{filename}",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()
```

### 1. Answers to the Core Problem Statement
1. **How Influence and Brokerage Appear in the Political Blog Network:**
   - Influence is heavily concentrated in a small oligarchy of elite blogs characterized by heavy-tailed, power-law in-degree and PageRank distributions.
   - Brokerage (Betweenness Centrality) is held by hyper-active campaign portals (e.g., `blogsforbush.com`) and prolific aggregators (`instapundit.com`) that link across factions and bridge peripheral blogs to core clusters.
2. **Whether Link Structure Recovers the Two Political Orientations:**
   - Without any access to political labels, unsupervised modularity maximization successfully partitions the hyperlink network into two major communities corresponding to the liberal and conservative wings with **88.54% pairwise accuracy, 91.71% precision, and 88.10% F1 score**. This confirms the presence of extreme political homophily and polarization in the 2004 blogosphere.
3. **How Central Blogs Affect Community Stability and Quality:**
   - The network displays classic scale-free vulnerability: removing just **2% (25 nodes)** of top betweenness brokers **fractures the network into 49 components**, spawns 56 fragmented communities, slashes recall by nearly 30%, and reduces F1 score to 0.6825. In contrast, equal-sized random removal leaves the network virtually intact (F1 = 0.8864). Elite brokers are the indispensable connective glue preserving the cohesion of political communities.



### 2. Critical Methodological Limitations & Historical Context
1. **Historical Specificity of the 2004 "Blogosphere":**
   - The 2004 U.S. election was a distinct historical era dominated by independent self-hosted weblogs, static "blogrolls," and manual permalinks. This differs fundamentally from modern algorithmic social media platforms (Twitter/X, Facebook, TikTok) driven by engagement algorithms, retweets, and algorithmic amplification.
2. **The Risk of Equating Hyperlinks with Endorsement:**
   - Treating hyperlinks as endorsements is a significant sociological oversimplification. In political argumentation, hyperlinks are frequently polemical and adversarial ("look what this opponent said"). A simple unweighted graph cannot distinguish between a positive endorsement and a scathing critique.
3. **Directional & Topological Simplification:**
   - Collapsing directed multigraphs into undirected simple graphs is mathematically necessary for standard Louvain modularity optimization, but discards the asymmetry between citing and being cited, as well as link frequency.
4. **Binary Ideological Assumption:**
   - The ground-truth classification reduces nuanced political discourse into a binary 0/1 dichotomy, erasing third-party perspectives, green politics, libertarianism, and independent moderates.
