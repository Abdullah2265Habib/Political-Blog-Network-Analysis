# Part A

### Loading the Dataset & Resolving the NetworkX GML Duplication Issue

```python
with open(RAW_DATA_PATH, "r", encoding="utf-8") as f: raw_content = f.read()

multi_content = raw_content.replace("graph [\n  directed 1", "graph [\n  directed 1\n  multigraph 1")
G_raw = nx.parse_gml(multi_content, label=None)

print("Raw Graph Data: ")
print(f"graph Class:              {type(G_raw).__name__}")
print(f"is Directed:              {G_raw.is_directed()}")
print(f"is Multigraph:            {G_raw.is_multigraph()}")
print(f"raw Node Count (|V|):     {G_raw.number_of_nodes():,}")
print(f"raw Edge Count (|E|):     {G_raw.number_of_edges():,}")
```

### Preprocessing the Audit Log

```python
missing_label = sum(1 for _, d in G_raw.nodes(data=True) if "label" not in d)
missing_val   = sum(1 for _, d in G_raw.nodes(data=True) if "value" not in d)
missing_src   = sum(1 for _, d in G_raw.nodes(data=True) if "source" not in d)
raw_val_dist  = Counter(d.get("value") for _, d in G_raw.nodes(data=True))

raw_self_loops = list(nx.selfloop_edges(G_raw))

edge_pair_counts = Counter((u, v) for u, v, _ in G_raw.edges(keys=True))
multi_pairs = {pair: count for pair, count in edge_pair_counts.items() if count > 1}
total_excess_edges = sum(c - 1 for c in multi_pairs.values())

raw_isolates = list(nx.isolates(G_raw))
raw_wccs = list(nx.weakly_connected_components(G_raw))
raw_sccs = list(nx.strongly_connected_components(G_raw))
wcc_sizes = sorted([len(c) for c in raw_wccs], reverse=True)
```

```python
print("Raw Preprocessing Audit")
print(f"Missing Node Labels:            {missing_label} (100% complete)")
print(f"Missing Political Values:       {missing_val} (Distribution: {dict(raw_val_dist)})")
print(f"Missing Source Attributes:      {missing_src}")
print(f"Self-loops Identified:          {len(raw_self_loops)} edges")
print(f"Blog Pairs with Multi-links:    {len(multi_pairs)} (Excess duplicate edges: {total_excess_edges})")
print(f"Isolated Nodes (Degree = 0):    {len(raw_isolates)} blogs")
print(f"Total Weakly Connected Comps:   {len(raw_wccs)} (Sizes: Giant={wcc_sizes[0]}, 2nd={wcc_sizes[1]}, Singletons={sum(1 for s in wcc_sizes if s == 1)})")
print(f"Total Strongly Connected Comps: {len(raw_sccs)} (Largest SCC size: {len(max(raw_sccs, key=len))})")
```
